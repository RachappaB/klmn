```python
import pandas as pd
import folium

# Load CSV
df = pd.read_csv('waling.csv')

# Drop rows where latitude or longitude is NaN
df = df.dropna(subset=['lat', 'lon'])

# Create a map centered at the average location
map_center = [df['lat'].mean(), df['lon'].mean()]
gps_map = folium.Map(location=map_center, zoom_start=12)

# Add GPS points to the map
for idx, row in df.iterrows():
    folium.Marker(
        location=[row['lat'], row['lon']],
        popup=f"Lat: {row['lat']}, Lon: {row['lon']}"
    ).add_to(gps_map)

# Save map to HTML
gps_map.save('gps_map.html')
print("Map saved as gps_map.html. Open it in a browser to view the GPS locations.")
```
