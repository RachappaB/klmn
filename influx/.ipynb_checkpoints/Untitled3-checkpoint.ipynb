```python
from influxdb_client import InfluxDBClient
import pandas as pd

# ================= CONFIG =================
INFLUXDB_URL = "https://us-east-1-1.aws.cloud2.influxdata.com"
INFLUXDB_TOKEN = "zoEJF1y0PDSeZuwoFZVOjr-k-9R2-vVdTCcug0_cIiArxYAiS4I1KlIYbl4eXPsavQB4Oppq1r0ONC3zogXVcg=="
ORG = "861c8c194327e7ec"
BUCKET = "habit"

# ================= CONNECT =================
client = InfluxDBClient(
    url=INFLUXDB_URL,
    token=INFLUXDB_TOKEN,
    org=ORG
)

query_api = client.query_api()

# ================= FLUX QUERY =================
query = f'''
from(bucket: "{BUCKET}")
  |> range(start: -1h)
  |> filter(fn: (r) => r._measurement == "iot_data")
  |> pivot(
      rowKey:["_time"],
      columnKey: ["_field"],
      valueColumn: "_value"
  )
  |> sort(columns: ["_time"])
'''

# ================= EXECUTE QUERY =================
df = query_api.query_data_frame(query)

# If multiple tables returned, merge them
if isinstance(df, list):
    df = pd.concat(df)

# ================= DISPLAY =================
print(df.head())
print("\nTotal rows:", len(df))

# ================= SAVE TO CSV =================
df.to_csv("iot_data.csv", index=False)
print("Saved to iot_data.csv")

client.close()
```
