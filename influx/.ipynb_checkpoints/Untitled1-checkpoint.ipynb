```python
query = f'''
from(bucket: "{BUCKET}")
  |> range(start: -10m)
  |> filter(fn: (r) => r._measurement == "iot_data")
'''

tables = query_api.query(query)
```

```python
print(tables)
```

```python
records = []

for table in tables:
    for row in table.records:
        records.append({
            "time": row.get_time(),
            "field": row.get_field(),
            "value": row.get_value()
        })

df_raw = pd.DataFrame(records)
```

```python
df = df_raw.pivot_table(
    index="time",
    columns="field",
    values="value"
).reset_index()

print(df.head())
```
