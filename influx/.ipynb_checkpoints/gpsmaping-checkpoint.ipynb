```python
from influxdb_client import InfluxDBClient
from influxdb_client.client.exceptions import InfluxDBError
import pandas as pd
import folium

# -------------------------------
# InfluxDB details
url = "https://us-east-1-1.aws.cloud2.influxdata.com"
org = "861c8c194327e7ec"
token="_7Ou6bUiKSraQ2Rhnwt_GVVQbFXtnU95_S-muf1U7lrzzKBTk2DjgiNXXrk332OZAslY3Mrqqk2g45YlKfyDBQ=="
bucket = "walking"
measurement = "health_data"
# -------------------------------

# Connect to InfluxDB
client = InfluxDBClient(url=url, token=token, org=org)

# Query all data (all-time)
query = f'''
from(bucket: "{bucket}")
  |> range(start: 0)
  |> filter(fn: (r) => r._measurement == "{measurement}")
'''

try:
    tables = client.query_api().query(query)
except InfluxDBError as e:
    print("InfluxDB query failed:", e)
    exit()

# Convert to a flat DataFrame
records = []
for table in tables:
    for record in table.records:
        records.append({
            "time": record.get_time(),
            "field": record.get_field(),
            "value": record.get_value()
        })

if not records:
    print("No data found in the bucket.")
    exit()

df = pd.DataFrame(records)

# Pivot so each field becomes a column
df = df.pivot(index='time', columns='field', values='value').reset_index()

# Optional: scale latitude and longitude if stored as int
if 'lat_scaled' in df.columns and 'lon_scaled' in df.columns:
    df['lat_scaled'] = df['lat_scaled'] / 1e6
    df['lon_scaled'] = df['lon_scaled'] / 1e6

# Drop rows with missing GPS data
if 'lat_scaled' in df.columns and 'lon_scaled' in df.columns:
    df = df.dropna(subset=['lat_scaled', 'lon_scaled'])

# Save CSV
df.to_csv("influx_data_all.csv", index=False)
print(f"Data downloaded successfully! Total rows: {len(df)}")

# -------------------------------
# GPS Map Visualization (if lat/lon exist)
if 'lat_scaled' in df.columns and 'lon_scaled' in df.columns:
    map_center = [df['lat_scaled'].mean(), df['lon_scaled'].mean()]
    gps_map = folium.Map(location=map_center, zoom_start=12)

    for _, row in df.iterrows():
        folium.Marker(
            location=[row['lat_scaled'], row['lon_scaled']],
            popup=f"Lat: {row['lat_scaled']}, Lon: {row['lon_scaled']}"
        ).add_to(gps_map)

    gps_map.save('gps_map_all.html')
    print("Map saved as gps_map_all.html. Open it in a browser to view the GPS locations.")
```

```python
# -------------------------------
# GPS Map Visualization (if lat/lon exist)
if 'lat_scaled' in df.columns and 'lon_scaled' in df.columns:
    map_center = [df['lat_scaled'].mean(), df['lon_scaled'].mean()]
    gps_map = folium.Map(location=map_center, zoom_start=14)

    # Create a list of all GPS points
    gps_points = df[['lat_scaled', 'lon_scaled']].values.tolist()

    # Add a PolyLine to connect all points
    folium.PolyLine(
        gps_points,
        color="blue",
        weight=3,
        opacity=0.7
    ).add_to(gps_map)

    # Optional: add a marker for start and end
    folium.Marker(gps_points[0], popup="Start", icon=folium.Icon(color='green')).add_to(gps_map)
    folium.Marker(gps_points[-1], popup="End", icon=folium.Icon(color='red')).add_to(gps_map)

    gps_map.save('gps_map_all.html')
    print("Map saved as gps_map_all.html with full path!")
```
